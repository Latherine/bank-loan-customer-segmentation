```python
import pandas as pd 
```

```python
#read cvs
df = pd.read_csv(r'C:\Users\ACER\Downloads\financial_loan.csv')
df.head()
```

```python
df['issue_date'] = pd.to_datetime(df['issue_date'], dayfirst=True, format ='mixed')
```

```python
df['month'] = df['issue_date'].dt.month_name()
```

```python
df.head()
```

```python
result = df.groupby('month').size().reset_index(name='count').sort_values('month')
print(result)
```
